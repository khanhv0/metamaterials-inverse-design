```python
import pandas as pd
from mesh_generator import MeshGenerator

design_parameters_list = pd.read_csv("./inverse_exp_results2.csv").iloc[:, :8].values
exp_df = pd.read_csv("./inverse_exp_results2.csv")
#count_ones = (exp_df.iloc[:, -1] == 1).sum()
#print(count_ones)
# List of the property prefixes
properties = ['poisson', 'young', 'volume_frac']

# Initialize a dictionary to store MAE values for each property
mae_values = {}

# Loop through each property and calculate MAE
for prop in properties:
    # Identify the requested and predicted columns for each property
    requested_col = f'{prop}_requested'
    predicted_col = f'{prop}_predicted'
    
    # Calculate MAE for each property
    mae = (exp_df[requested_col] - exp_df[predicted_col]).abs().mean()
    
    # Store the result in the dictionary
    mae_values[prop] = mae

# Print the MAE results for each property
print("Mean Absolute Error (MAE) for each property:")
for prop, mae in mae_values.items():
    print(f"{prop.capitalize()} MAE: {mae:.4f}")
last_column = exp_df.iloc[:, -1]  # Access the last column
count_ones = (last_column == 1).sum()  # Count where the value is 1
print(count_ones)
# mesh_generator = MeshGenerator()
# results = mesh_generator.generateMeshesBatch(design_parameters_list, plot = True, output_path="inverse_model_exp_validation")
```
